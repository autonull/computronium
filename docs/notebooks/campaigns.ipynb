# Running Campaigns Tutorial

Campaigns allow you to run systematic comparisons across multiple architectures, seeds, and configurations.

## Using the Campaign CLI

The `comp campaign` command runs declarative multi-run campaigns from YAML configurations.

```python
# View campaign help
!comp campaign --help
```

## Creating a Campaign YAML

Let's create a campaign that compares different credit assignment methods:

```python
campaign_yaml = """
name: credit_comparison
description: Compare credit assignment methods on MNIST

models:
  - name: backprop
    coordinate: digital/feedforward/instantaneous/null/gradient/euclidean
  - name: equilibrium_prop
    coordinate: digital/recurrent/energy_minimization/null/equilibrium_prop/adam
  - name: feedback_alignment
    coordinate: digital/feedforward/instantaneous/null/feedback_alignment/euclidean
  - name: pepita
    coordinate: digital/recurrent/energy_minimization/null/pepita/adam

tasks:
  - digits

seeds: [42, 123, 456]
epochs: 20
batch_size: 128

objectives:
  - validation_accuracy
  - walltime_total
  - energy_per_step
  - spectral_radius
  - settle_steps

output:
  store: credit_comparison.db
  report: credit_comparison.html
"""

with pathlib.Path("credit_campaign.yaml").open("w") as f:
    f.write(campaign_yaml)

print("Campaign YAML created: credit_campaign.yaml")
```

## Running the Campaign

```python
# Run the campaign (this may take a few minutes)
# Uncomment to run:
# !comp campaign --config credit_campaign.yaml
```
