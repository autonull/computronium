## What is Computronium?

Computronium is an experimental platform for exploring the 6-dimensional space of neural network architectures:

1. **Substrate** - Physical implementation (digital, analog, memristive, photonic, neuromorphic, quantum)
2. **Geometry** - Network topology (feedforward, recurrent, transformer, attention, mixture-of-experts)
3. **Dynamics** - State evolution (instantaneous, energy_minimization, predictive_settling, spike_integration)
4. **Plasticity** - Weight adaptation (null, routing, fast_weights, rule_state, substrate_coupled, temporal_psi)
5. **Credit** - Credit assignment (gradient, equilibrium_prop, feedback_alignment, pepita, target_prop)
6. **Update** - Parameter update rule (euclidean, adam, lion, muon, riemannian, spectral)

## Running a Simple Experiment

```python
# Define a simple coordinate (6-D architecture specification)
coordinate = "digital/feedforward/instantaneous/null/gradient/euclidean"

# This represents:
# - Substrate: digital (standard PyTorch)
# - Geometry: feedforward (MLP)
# - Dynamics: instantaneous (no settling)
# - Plasticity: null (no weight adaptation)
# - Credit: gradient (backpropagation)
# - Update: euclidean (standard SGD)
print(f"Coordinate: {coordinate}")
```

```python
# Run the MNIST benchmark with a simple MLP
# First, let's create a model manually
from torch import nn


class SimpleMLP(nn.Module):
    def __init__(self, input_dim=784, hidden_dim=256, output_dim=10):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, output_dim),
        )

    def forward(self, x):
        return self.net(x.view(x.size(0), -1))


model = SimpleMLP().to(device)
print(f"Model parameters: {sum(p.numel() for p in model.parameters()):,}")
```

```python
# Run the MNIST benchmark
result = mnist_benchmark(model)

print(f"Validation Accuracy: {result.metrics.get('validation_accuracy', 'N/A'):.4f}")
print(f"Validation Loss: {result.metrics.get('validation_loss', 'N/A'):.4f}")
print(f"Training Time: {result.metadata.get('training_time_seconds', 'N/A'):.2f}s")
```
